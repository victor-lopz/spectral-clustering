### Plot trajectories for the Duffing dynamical system

```python
import numpy as np

from spectral_clustering import (
    SpectralClusteringConfig,
    ode_autonomous_duffing,
    ode_non_autonomous_duffing,
    plot_trajectory_paths,
    simulate_trajectories,
)

%load_ext autoreload
%autoreload 2
```

```python
params = SpectralClusteringConfig(
    num_dimensions=2,
    t_span=(0, 22),
    t_steps=2000,
    x_min=-1,
    x_max=1,
    y_min=-0.5,
    y_max=0.5,
)
initial_conditions = np.array([[-0.85, -0.65], [0.55, 0.65], [0.2, 0.0]])
# initial_conditions = np.array([[-1.0, -0.75]])
```

#### Non-autonomous Duffing system

```python
trajectories = simulate_trajectories(
    ode_non_autonomous_duffing, initial_conditions, params
)
plot_trajectory_paths(
    trajectories, subfolder="non_autonomous", plot_title="Non autonomous Duffing"
)
```

#### Autonomous Duffing system

```python
trajectories = simulate_trajectories(ode_autonomous_duffing, initial_conditions, params)
plot_trajectory_paths(
    trajectories, subfolder="autonomous", plot_title="Autonomous Duffing"
)
```
